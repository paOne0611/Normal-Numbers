## Simplifying f_4(x,y)

```python
import sympy as sp
```

```python
x, y, s= sp.symbols('x y s')  ##variables in the expr
n, k = sp.symbols('n k', integer=True, nonnegative=True) ## naturals
p = sp.Symbol('p', integer=True)  ##running index
```

```python
term= (
    sp.binomial(n,p)
    * (((s+1)*p-n)**k)
    * (x**(s*p))
    * (y**(n-p))
    )
```

```python
f_k= sp.Sum(term, (p, 0, n))
```

```python
display(f_k)
```

```python
f_4= sp.Sum(
    (
        sp.binomial(n,p)
        * (((s+1)*p-n)**4)
        * (x**(s*p))
        * (y**(n-p))
    ), (p, 0, n))
    
```

```python
result= sp.expand(f_4)
result
```

## Another approach

```python
f0 = (x**s + y)**n
```

```python
def D(expr):
    return (s+1)*x*sp.diff(expr, x) - n*expr
```

```python
f4 = f0
for _ in range(4):
    f4 = sp.expand(D(f4))  ###recursively evaluating D(expr)
```

```python
sp.factor(f4)
```
